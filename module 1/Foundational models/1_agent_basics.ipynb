# LangChain Agent Basics: Models, Dials, and System Prompts

This notebook covers:
1. Initializing a model and inspecting the full response object (including metadata & tokens).
2. Experimenting with temperature settings (Low vs. High).
3. Using `create_react_agent` with a System Prompt and Few-Shot Examples.

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
from langgraph.prebuilt import create_react_agent

# Load environment variables (GROQ_API_KEY)
load_dotenv()
```

```python
# Initialize a fast, cost-effective LLM
llm = ChatGroq(
    model_name="qwen/qwen3.8-27b",
    temperature=0.7,
    max_tokens=200
)

# Invoke the model
response = llm.invoke("What is an AI agent in 2 sentences?")

print("--- FULL RESPONSE OBJECT ---")
print(response)

print("\n--- RESPONSE METADATA & TOKEN USAGE ---")
print("Token Usage:", response.response_metadata.get("token_usage"))
print("Model Name:", response.response_metadata.get("model_name"))
print("Finish Reason:", response.response_metadata.get("finish_reason"))

print("\n--- CONTENT ONLY ---")
print(response.content)
```

```python
from langchain_core.messages import HumanMessage
from langchain.agents import create_agent

# System prompt shaping behavior with few-shot examples
system_prompt = """You are a strict, ultra-concise tech jargon translator.
Translate complex tech terms into 1-word analogies for 5-year-olds.

Examples:
User: What is an API?
Assistant: Doorbell

User: What is a Database?
Assistant: Toybox

User: What is a Server?
Assistant: Kitchen
"""

# Initialize Agent using system_prompt
agent = create_agent(
    model=llm,
    tools=[],
    system_prompt=system_prompt  # <-- Changé 'prompt' par 'system_prompt'
)

# Run Agent
agent_response = agent.invoke({
    "messages": [HumanMessage(content="What is Kubernetes?")]
})

print("--- AGENT CONVERSATION TRACE ---")
for msg in agent_response["messages"]:
    print(f"[{msg.type.upper()}]: {msg.content}")
```

```python
from langchain_core.messages import HumanMessage
from langchain.agents import create_agent

system_prompt = """You are a lazy, sarcastic agent. 
Respond to every technical query with EXACTLY 2 emojis that represent the concept. Do not use any text.

Examples:
User: What is an API?
Assistant: 🚪🔔

User: What is a Database?
Assistant: 📦💾

User: What is a Server?
Assistant: 🖥️⚡
"""

agent = create_agent(
    model=llm,
    tools=[],
    system_prompt=system_prompt
)

agent_response = agent.invoke({
    "messages": [HumanMessage(content="What is Cloud Computing?")]
})

print("--- AGENT CONVERSATION TRACE ---")
for msg in agent_response["messages"]:
    print(f"[{msg.type.upper()}]: {msg.content}")
```
